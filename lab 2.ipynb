```python
corr = df.corr(numeric_only=True)

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 8))
plt.imshow(corr)

plt.colorbar()

plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)

plt.title("Correlation Matrix")

plt.show()
```

```python
plt.scatter(df['diabetes'], df['bmi'])

plt.xlabel('Diabetes')
plt.ylabel('BMI')
plt.title('Diabetes vs BMI')

plt.show()
```

```python
plt.plot(df['age'], df['blood_glucose_level'])

plt.xlabel('Age')
plt.ylabel('Blood Glucose Level')
plt.title('Age vs Blood Glucose Level')

plt.show()
```

```python
import matplotlib.pyplot as plt

count = df['diabetes'].value_counts()

plt.bar(count.index, count.values)

plt.xlabel('Diabetes')
plt.ylabel('Count')
plt.title('Diabetes Count')

plt.show()
```

```python
import matplotlib.pyplot as plt

plt.hist(df['bmi'])

plt.xlabel('BMI')
plt.ylabel('Frequency')
plt.title('BMI Distribution')

plt.show()
```

```python
import matplotlib.pyplot as plt

plt.boxplot(df['bmi'])

plt.ylabel('BMI')
plt.title('BMI Box Plot')

plt.show()
```

```python
import matplotlib.pyplot as plt

count = df['diabetes'].value_counts()

plt.pie(count.values, labels=count.index, autopct='%1.1f%%')

plt.title('Diabetes Distribution')

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.countplot(x='gender', data=df)

plt.xlabel('Gender')
plt.ylabel('Count')
plt.title('Gender Count')

plt.show()
```
